```python
# Contoh soal:
#  2x +  y -  z =   8
# -3x -  y + 2z = -11
# -2x +  y + 2z =  -3
A = np.array([[ 2,  1, -1],
              [-3, -1,  2],
              [-2,  1,  2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)

print("Matriks A:\n", A)
print("Vektor b :", b)
```

Penjelasan:

A adalah matriks koefisien 3×3, satu baris untuk satu persamaan.
b adalah vektor ruas kanan.
dtype=float mencegah pembagian terpotong menjadi bilangan bulat saat eliminasi.
print menampilkan sistem awal agar bisa dicek.

```python
def eliminasi_gauss(A, b, tampil=True):
    n = len(b)                                   # jumlah persamaan (3)
    M = np.hstack([A.copy(), b.reshape(-1, 1)])  # matriks augmented [A|b]

    # TAHAP 1: ELIMINASI MAJU
    for k in range(n - 1):                       # k = kolom pivot
        p = np.argmax(np.abs(M[k:, k])) + k      # baris dengan |nilai| terbesar
        if abs(M[p, k]) < 1e-12:                 # pivot ~ 0 berarti singular
            raise ValueError("Matriks singular: solusi tunggal tidak ada.")
        if p != k:
            M[[k, p]] = M[[p, k]]                # tukar baris k dan p

        for i in range(k + 1, n):                # nolkan elemen di bawah pivot
            faktor = M[i, k] / M[k, k]
            M[i, k:] = M[i, k:] - faktor * M[k, k:]

        if tampil:
            print(f"Setelah eliminasi kolom {k+1}:\n{np.round(M, 4)}\n")

    if abs(M[n-1, n-1]) < 1e-12:                 # cek pivot terakhir
        raise ValueError("Matriks singular: solusi tunggal tidak ada.")

    # TAHAP 2: SUBSTITUSI MUNDUR
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):               # dari baris bawah ke atas
        x[i] = (M[i, -1] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]
    return x
```

Penjelasan: Ini inti algoritma.

np.hstack menempelkan b sebagai kolom terakhir A sehingga terbentuk matriks augmented. A.copy() menjaga data asli tidak berubah.
Pivoting parsial: di tiap kolom dicari elemen terbesar (nilai mutlak) lalu barisnya ditukar ke posisi pivot. Ini mengurangi galat pembulatan dan menghindari pembagian dengan nol.
Jika pivot terbesar mendekati nol (< 1e-12), sistem singular dan fungsi berhenti dengan pesan error.
faktor = M[i,k]/M[k,k] adalah pengali eliminasi. Baris i dikurangi faktor × baris pivot, sehingga elemen di bawah pivot menjadi nol. Hasilnya matriks segitiga atas.
Substitusi mundur menghitung variabel terakhir (z) lebih dulu, lalu y dan x. Rumusnya x_i = (b_i − Σ a_ij·x_j) / a_ii, dengan np.dot menghitung jumlahnya.

```python
solusi = eliminasi_gauss(A, b)

print("HASIL:")
print(f"x = {solusi[0]:.4f}")
print(f"y = {solusi[1]:.4f}")
print(f"z = {solusi[2]:.4f}")
```

Penjelasan: Fungsi dipanggil dengan A dan b dari cell 2. Output berupa matriks tiap tahap eliminasi, lalu hasil akhir dengan 4 desimal. Hasil yang benar adalah x=2, y=3, z=−1.

```python
pembanding = np.linalg.solve(A, b)
print("Solusi numpy   :", np.round(pembanding, 4))

residual = A @ solusi - b
print("Residual A.x-b :", residual)

print("Cocok dengan numpy?", np.allclose(solusi, pembanding))
```

Penjelasan:

np.linalg.solve adalah solver bawaan numpy sebagai pembanding.
Residual A·x − b seharusnya mendekati 0. Operator @ adalah perkalian matriks.
np.allclose bernilai True jika dua hasil sama dalam toleransi kecil.

```python
print("Masukkan koefisien a1 a2 a3 dan konstanta b (pisahkan spasi).")
A_in, b_in = [], []
for i in range(3):
    baris = list(map(float, input(f"Persamaan {i+1}: ").split()))
    A_in.append(baris[:3])
    b_in.append(baris[3])

hasil = eliminasi_gauss(np.array(A_in), np.array(b_in), tampil=False)
print("\nx, y, z =", np.round(hasil, 4))
```

Penjelasan:

input() membaca satu baris teks, split() memecahnya per spasi, dan map(float, ...) mengubahnya menjadi angka.
Tiga angka pertama menjadi koefisien dan angka keempat menjadi konstanta. Contoh input untuk 2x+y−z=8 adalah 2 1 -1 8.
tampil=False menyembunyikan langkah eliminasi agar output ringkas.
